```python
# 1. Load the first file (MnS-1)
df = pd.read_csv(xrd_file + 'MnS -1-GIXRD omega05deg.xy', sep=r'\s+', header=None)

# 2. Normalize intensity (0 to 1) just like in the stack plot
df[1] = (df[1] - df[1].min()) / (df[1].max() - df[1].min())

# 3. Crop tightly around the hump (28° to 40°)
crop = df[(df[0] >= 28) & (df[0] <= 40)]
x = crop[0].values
y = crop[1].values

# 4. Plot to confirm the cropped hump
plt.figure(figsize=(7, 3.5))
plt.plot(x, y, 'b.-', label='MnS-1 Cropped Hump')
plt.xlabel('2-Theta (°)')
plt.ylabel('Normalized Intensity')
plt.title('Cropped Region (28° - 40°)')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

```python
from scipy.optimize import curve_fit

# 1. Model: Gaussian peak + linear background slope
def model(x, a, x0, sigma, m, c):
    return a * np.exp(-((x - x0)**2) / (2 * sigma**2)) + (m * x + c)

# 2. Fit to cropped data
p0 = [y.max() - y.min(), 33.5, 1.5, -0.05, y[0]]
popt, _ = curve_fit(model, x, y, p0=p0)
a, x0, sigma, m, c = popt

# 3. Peak Center & FWHM
fwhm = 2.355 * abs(sigma)

print(f"Peak Center: {x0:.2f}°")
print(f"FWHM: {fwhm:.2f}°")
```

```python
# Plot data, fitted curve, and background line
plt.figure(figsize=(7, 3.5))
plt.plot(x, y, 'b.', label='Data')
plt.plot(x, model(x, *popt), 'r-', label=f'Fit (Peak={x0:.1f}°)')
plt.plot(x, m * x + c, 'k--', alpha=0.5, label='Background')
plt.xlabel('2-Theta (°)')
plt.ylabel('Normalized Intensity')
plt.legend()
plt.show()
```

```python
# Scherrer Equation: D = (K * lambda) / (beta * cos(theta))
K = 0.9                   # Shape factor
wavelength = 0.15406      # Cu K-alpha X-ray wavelength (nm)

theta_rad = np.radians(x0 / 2)   # Half angle in radians
beta_rad = np.radians(fwhm)      # FWHM in radians

D = (K * wavelength) / (beta_rad * np.cos(theta_rad))

print(f"Crystallite Domain Size (D): {D:.2f} nm")
```

```python
from scipy.signal import savgol_filter

# 1. Smooth the blue curve to kill the tiny 0.1° noise wiggles
y_smooth = savgol_filter(df[1], window_length=31, polyorder=3)

# 2. Find only the TRUE broad peaks (width > 20 points)
peaks, props = find_peaks(y_smooth, distance=50, prominence=0.02, width=15)

# 3. Plot to verify
plt.figure(figsize=(9, 4))
plt.plot(df[0], df[1], alpha=0.3, label='Raw Noisy Data')
plt.plot(df[0], y_smooth, 'b-', label='Real Smoothed Curve')
plt.plot(df[0].iloc[peaks], y_smooth[peaks], 'ro', label='True Broad Features')
plt.xlabel('2-Theta (°)')
plt.ylabel('Normalized Intensity')
plt.legend()
plt.show()
```

```python
print("Feature | 2θ (°) | FWHM (°) | Crystallite Size (nm) | Assignment")
print("-" * 65)

step = df[0].iloc[1] - df[0].iloc[0]  # scan step size

for i, p in enumerate(peaks):
    two_theta = df[0].iloc[p]
    fwhm = props['widths'][i] * step
    
    # Scherrer Equation
    theta = np.radians(two_theta / 2)
    beta = np.radians(fwhm)
    size = (0.9 * 0.15406) / (beta * np.cos(theta))
    
    # Identify what it actually is
    desc = "Glass Substrate" if two_theta < 26 else ("MnS Main Phase" if two_theta < 36 else "MnS (220)")
    
    print(f"Peak {i+1}  | {two_theta:5.1f}° | {fwhm:5.2f}° | {size:6.2f} nm        | {desc}")
```

```python
from scipy.optimize import curve_fit

# Use column 0 (angle) and column 1 (intensity)
x = crop[0].values
y = crop[1].values

# 1. Model: Bell curve (Gaussian) + Sloping straight line (Background)
def model(x, a, x0, sigma, m, c):
    peak = a * np.exp(-((x - x0) ** 2) / (2 * sigma ** 2))
    background = m * x + c
    return peak + background

# 2. Starting guess: [amplitude, peak_center, width, slope, intercept]
p0 = [y.max() - y.min(), 33.5, 1.5, -0.05, y[0]]

# 3. Fit the model to the data
popt, _ = curve_fit(model, x, y, p0=p0)
a, x0, sigma, m, c = popt

# 4. Calculate FWHM (Full Width at Half Maximum)
fwhm = 2.355 * abs(sigma)

print(f"Fit Complete!")
print(f"Peak Center (2θ) : {x0:.2f}°")
print(f"Peak Width (FWHM): {fwhm:.2f}°")
```

```python
xrd_file = '/kaggle/working/MnS manuscript dataset/XRD/'
plt.figure(figsize=(10, 8))
offset = 0.0

for file in sorted(os.listdir(xrd_file)):
    df = pd.read_csv(xrd_file + file, sep=r'\s+', header=None)
    x = df[0].values
    y = df[1].values
    
    # --- Step 1: Baseline Correction (Remove curved glass background) ---
    baseline = np.polyval(np.polyfit(x, y, deg=3), x)
    y_corrected = np.maximum(y - baseline, 0)
    
    # --- Step 2: Normalize (0 to 1) & Stack ---
    y_norm = y_corrected / y_corrected.max()
    plt.plot(x, y_norm + offset, label=file[:12])
    offset += 0.5  # Vertical shift for stacking

plt.xlabel('2-Theta (°)')
plt.ylabel('Baseline-Corrected Intensity (Shifted)')
plt.title('Baseline-Corrected & Stacked GIXRD Patterns')
plt.legend(loc='upper right', fontsize=8)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 6))

for file in sorted(os.listdir(xrd_file)):
    df = pd.read_csv(xrd_file + file, sep=r'\s+', header=None)
    x = df[0]
    y = df[1] / df[1].max()
    
    # 1. Fit a simple polynomial line to estimate the glass background
    background = np.polyval(np.polyfit(x, y, deg=3), x)
    
    # 2. Subtract background to get a flat curve
    y_flat = y - background
    y_flat = np.maximum(y_flat, 0)  # remove negative values
    
    plt.plot(x, y_flat, label=file[:12])

plt.xlabel('2-Theta (degrees)')
plt.ylabel('Background-Subtracted Intensity')
plt.title('XRD - Flattened Patterns (Glass Background Removed)')
plt.legend(fontsize=7)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
afm_dir = '/kaggle/working/MnS manuscript dataset/AFM-scans/'
files = sorted(f for f in os.listdir(afm_dir) if f.endswith('.png'))

# 1. Calculate Ra and Rq for every AFM file
names, Ra_vals, Rq_vals = [], [], []

for f in files:
    h = np.array(Image.open(afm_dir + f).convert('L')).astype(float)
    h_flat = h - h.mean()
    Ra_vals.append(np.mean(np.abs(h_flat)))
    Rq_vals.append(np.sqrt(np.mean(h_flat**2)))
    names.append(f[:18])

# 2. Bar chart comparison
x = np.arange(len(names))
fig, ax = plt.subplots(figsize=(16, 5))
ax.bar(x - 0.2, Ra_vals, 0.4, label='Ra (Avg Roughness)', color='teal')
ax.bar(x + 0.2, Rq_vals, 0.4, label='Rq (RMS Roughness)', color='coral')

ax.set_xticks(x)
ax.set_xticklabels(names, rotation=90, fontsize=7)
ax.set_ylabel('Roughness (a.u.)')
ax.set_title('AFM Surface Roughness Comparison — All Samples')
ax.legend()

plt.tight_layout()
plt.savefig('/kaggle/working/afm_roughness_comparison.png', dpi=150)
plt.show()
```

```python
import os, numpy as np
from PIL import Image
import plotly.graph_objects as go
from plotly.subplots import make_subplots

afm_dir = '/kaggle/working/MnS manuscript dataset/AFM-scans/'

# Pick key T150 comparison files
key_files = [f for f in sorted(os.listdir(afm_dir)) if 'T150' in f and '1ku1' in f and '1M' in f and f.endswith('.png')]

# Build interactive subplots
fig = make_subplots(rows=1, cols=len(key_files), specs=[[{'type': 'surface'}] * len(key_files)],
                    subplot_titles=[f[:18] for f in key_files])

for i, f in enumerate(key_files, 1):
    h = np.array(Image.open(afm_dir + f).convert('L')).astype(float)
    x = np.linspace(0, 2.5, h.shape[1])
    y = np.linspace(0, 2.5, h.shape[0])
    fig.add_trace(go.Surface(z=h, x=x, y=y, colorscale='hot', showscale=False), row=1, col=i)

fig.update_layout(title='Interactive AFM 3D Surfaces (T150, 1:1, 1M)', height=600)
fig.show()
```

```python
import numpy as np, matplotlib.pyplot as plt
from PIL import Image

# 1. Load height map
h = np.array(Image.open('/kaggle/working/MnS manuscript dataset/AFM-scans/MnS PA 1ku1 c1 T150 2,5x2,5.png').convert('L')).astype(float)

# 2. Extract horizontal line profile at the middle row
mid_row = h.shape[0] // 2
line_profile = h[mid_row, :]
distance = np.linspace(0, 2.5, len(line_profile))  # in µm

# 3. Plot Image + Line Profile side by side
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 4))

ax1.imshow(h, cmap='afmhot')
ax1.axhline(mid_row, color='cyan', lw=1.5, label='Profile Line')
ax1.set_title('2D Topography + Profile Line')
ax1.legend(); ax1.axis('off')

ax2.plot(distance, line_profile, color='teal')
ax2.set_xlabel('Distance (µm)')
ax2.set_ylabel('Height (a.u.)')
ax2.set_title('Cross-Sectional Height Profile')
ax2.grid(True, alpha=0.3)

plt.tight_layout()
plt.savefig('/kaggle/working/afm_line_profile.png', dpi=150)
plt.show()
```
